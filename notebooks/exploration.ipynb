```python
import pandas as pd
import geopandas as gpd
import folium
from branca.colormap import linear

# Load street network segments
streets = gpd.read_file("../data/raw/nyc_street_network.geojson")

# Load aggregated crash counts by segment-day
crashes = pd.read_csv("../data/processed/segment_daily_crashes.csv")

# Aggregate to total crashes per segment across all time
segment_totals = (
    crashes.groupby("segment_id", as_index=False)["crash_count"]
    .sum()
    .rename(columns={"crash_count": "total_crashes"})
)

# Rebuild segment_id to match your pipeline
streets = streets.reset_index(drop=True).copy()
streets["segment_id"] = streets.index.astype(int)

# Merge crash totals onto street segments
gdf = streets.merge(segment_totals, on="segment_id", how="left")
gdf["total_crashes"] = gdf["total_crashes"].fillna(0)

# Keep only segments with at least 1 crash for readability
gdf_plot = gdf[gdf["total_crashes"] > 0].copy()

# Reproject to WGS84 for folium
gdf_plot = gdf_plot.to_crs(epsg=4326)

print("Segments with crashes:", len(gdf_plot))
print(gdf_plot[["segment_id", "total_crashes"]].head())

# Build map
m = folium.Map(location=[40.7128, -74.0060], zoom_start=11, tiles="cartodbpositron")

# Color scale
max_crashes = gdf_plot["total_crashes"].max()
colormap = linear.YlOrRd_09.scale(1, max_crashes)
colormap.caption = "Total Crashes per Segment"
colormap.add_to(m)

# Style function
def style_function(feature):
    crashes = feature["properties"]["total_crashes"]
    return {
        "color": colormap(crashes),
        "weight": 2.5,
        "opacity": 0.8,
    }

# Tooltip
tooltip = folium.GeoJsonTooltip(
    fields=["segment_id", "total_crashes"],
    aliases=["Segment ID", "Total Crashes"],
    sticky=False
)

# Add segments
folium.GeoJson(
    gdf_plot[["segment_id", "total_crashes", "geometry"]].to_json(),
    style_function=style_function,
    tooltip=tooltip
).add_to(m)

m.save("segment_crash_map.html")
```
